```python
pdf = pdfplumber.open(file_path)

for page in pdf.pages:
# p0 = pdf.pages[4]
    im = page.to_image()
# im.debug_tablefinder(table_settings = table_setting)
# im.debug_tablefinder(table_settings = table_setting2)
    im.draw_rects(page.extract_words(use_text_flow=True, x_tolerance=3, extra_attrs=["fontname"]))
    im.show()
# p0.extract_text(use_text_flow=True, x_tolerance=1, layout=True)
```

```python
pdf = pdfplumber.open(file_path)

page_data = {
    "page": 0,
    "text_blocks": [],
    "images": [],
    "tables": []
}

page = pdf.pages[0]
im = page.to_image()
# im.debug_tablefinder(table_settings = table_setting)
# im.debug_tablefinder(table_settings = table_setting2)
words = page.extract_words(use_text_flow=False, x_tolerance=4, extra_attrs=["fontname"])
im.draw_rects(page.extract_words(use_text_flow=False, x_tolerance=4, extra_attrs=["fontname"]))
im.show()

for w in words:
    page_data["text_blocks"].append({
        "text": w["text"],
        "bbox": (w["x0"], w["top"], w["x1"], w["bottom"])
    })

print(f"\n\n{page_data["text_blocks"]}\n\n")

def page_to_text_with_images(page: dict) -> str:
    """
    Combine text blocks, OCR from images, and table text into a single string.
    """
    text = " ".join([tb["text"] for tb in page["text_blocks"]])
    ocr_texts = " ".join([img["ocr_text"] for img in page["images"]])    
    tables_text = " ".join([tbl["text"] for tbl in page.get("tables", [])])
    

    return [text, ocr_texts, tables_text]

print(page_to_text_with_images(page_data))
# p0.extract_text(use_text_flow=True, x_tolerance=1, layout=True)
```
